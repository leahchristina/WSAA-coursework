This program retrieves the Exchequer Account (Historial Series) dataset from the Irish CSO API and saves the returned JSON data to a local file.

```python
import requests
import json
```

```python
url = "https://ws.cso.ie/public/api.restful/PxStat.Data.Cube_API.ReadDataset/FIQ02/JSON-stat/1.0/en"

response = requests.get(url)

with open("cso.json", "w") as file:
    json.dump(response.json(), file, indent=4)

print("Data saved to cso.json")
```
